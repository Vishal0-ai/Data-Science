# Operators


```python
#Arithmetic Operators

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b) # Give Floating Value
print(a // b) # Give Int Value
print(a % b)
print(a ** b)
```

```python
# Comparison Operators

a = 10
b = 20

print(a == b)  
print(a != b)  
print(a > b)   
print(a < b)   
```
