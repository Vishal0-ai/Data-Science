## Tuples Methods

```python
# Tuple Methods Practice Program

numbers = (40, 10, 30, 20, 10, 50, 10)

print("Original Tuple:", numbers)

# 1. count()
count = numbers.count(10)
print("Count of 10:", count)

# 2. index()
position = numbers.index(30)
print("Index of 30:", position)

# 3. Accessing elements
print("First element:", numbers[0])
print("Last element:", numbers[-1])

# 4. Slicing
print("Sliced Tuple:", numbers[1:5])

# 5. Reverse using slicing
print("Reversed Tuple:", numbers[::-1])

# 6. Length
print("Length:", len(numbers))

# 7. Maximum
print("Maximum:", max(numbers))

# 8. Minimum
print("Minimum:", min(numbers))

# 9. Sum
print("Sum:", sum(numbers))

# 10. Check element
print("20 exists:", 20 in numbers)

# 11. Concatenation
another_tuple = (60, 70)

combined = numbers + another_tuple

print("After concatenation:", combined)

# 12. Repetition
repeated = (1, 2) * 3

print("Repeated Tuple:", repeated)
```
