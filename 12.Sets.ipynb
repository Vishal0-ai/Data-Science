```python
numbers = {10, 20, 20, 30, 30}

print(numbers)
```

## Methods

```python
# Set Methods Practice Program

numbers = {10, 20, 30, 40, 50}

print("Original Set:", numbers)

# 1. add()
numbers.add(60)
print("After add():", numbers)

# 2. update()
numbers.update([70, 80, 90])
print("After update():", numbers)

# 3. remove()
numbers.remove(20)
print("After remove():", numbers)

# 4. discard()
numbers.discard(30)
print("After discard():", numbers)

# 5. pop()
removed = numbers.pop()
print("Element removed by pop():", removed)
print("After pop():", numbers)

# 6. copy()
new_numbers = numbers.copy()
print("Copied Set:", new_numbers)

# 7. union()
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print("Union:", set1.union(set2))

# 8. intersection()
print("Intersection:", set1.intersection(set2))

# 9. difference()
print("Difference:", set1.difference(set2))

# 10. symmetric_difference()
print("Symmetric Difference:",
      set1.symmetric_difference(set2))

# 11. issubset()
small_set = {1, 2}

print("Is subset:", small_set.issubset(set1))

# 12. issuperset()
print("Is superset:", set1.issuperset(small_set))

# 13. isdisjoint()
set3 = {7, 8, 9}

print("Is disjoint:", set1.isdisjoint(set3))

# 14. clear()
numbers.clear()
print("After clear():", numbers)
```

## Operators

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print(set1 | set2)  # Union
print(set1 & set2)  # Intersection
print(set1 - set2)  # Difference
print(set1 ^ set2)  # Symmetric difference
```
