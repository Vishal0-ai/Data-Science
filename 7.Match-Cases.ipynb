# Match Cases

```python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
```

```python
def calculator(a, b, operation):

    match operation:
        case "+":
            return a + b
        case "-":
            return a - b
        case "*":
            return a * b
        case "/":
            return a / b
        case _:
            return "Invalid operation"


print(calculator(10, 5, "+"))
print(calculator(10, 5, "*"))
```

```python
choice = int(input("Enter your choice: "))

match choice:
    case 1:
        print("Add")
    case 2:
        print("Update")
    case 3:
        print("Delete")
    case 4:
        print("Exit")
    case _:
        print("Invalid choice")
```
