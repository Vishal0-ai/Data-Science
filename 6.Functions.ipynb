# Functions

```python
def greet():
    print("Hello!")
```

```python
greet()
```

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(60, 40)
```

```python
def check_age(age):
    if age >= 18:
        return "Adult"
    else:
        return "Minor"

result = check_age(22)

print(result)
```

```python
def greet(name="Guest"):
    print("Hello", name)

greet("Vishal")
greet()
```

```python
def check_result(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "Fail"


marks = int(input("Enter your marks: "))

grade = check_result(marks)

print("Your grade is:", grade)
```
