# Loops in python


# For Loops

```python
for i in range(5):
    print(i)
```

```python
for i in range(2, 6):
    print(i)
```

```python
name = "Vishal"

for char in name:
    print(char)
```

```python
student = {
    "name": "Amit",
    "age": 22,
    "city": "Delhi"
}

for key in student:
    print(key)
```

## While loop

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# Break 
```

```python
for i in range(10):
    if i == 5:
        break

    print(i)
```

### Nested Loop

```python
for i in range(1, 5):
    for j in range(i):
        print("*", end=" ")

    print()
```
