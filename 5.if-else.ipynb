# If-Else


```python
age = 20

if age >= 18:
    print("You are an adult")
```

```python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

```python
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

```python
age = 25

if age >= 18 and age <= 60:
    print("Eligible")
```

```python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("Weekend")
```
