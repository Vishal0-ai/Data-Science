```python
# A Complete Program
import os

filename = "students.txt"

# Write
with open(filename, "w") as file:
    file.write("Amit\n")
    file.write("Priya\n")
    file.write("Rahul\n")

print("Data written successfully.")

# Read
with open(filename, "r") as file:
    data = file.read()

print("\nFile content:")
print(data)

# Append
with open(filename, "a") as file:
    file.write("Vishal\n")

print("Data appended successfully.")

# Read line by line
print("\nUpdated content:")

with open(filename, "r") as file:
    for line in file:
        print(line.strip())

# Check file
if os.path.exists(filename):
    print("\nFile exists.")

# Delete
# os.remove(filename)
```
