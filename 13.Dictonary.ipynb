## Dictionary Methods

```python
# Dictionary Methods Practice Program

student = {
    "name": "Vishal",
    "age": 22,
    "course": "B.Tech CSE",
    "city": "Delhi"
}

print("Original Dictionary:", student)

# 1. get()
print("Name:", student.get("name"))

# 2. keys()
print("Keys:", student.keys())

# 3. values()
print("Values:", student.values())

# 4. items()
print("Items:", student.items())

# 5. update()
student.update({
    "age": 23,
    "college": "Galgotias University"
})

print("After update():", student)

# 6. setdefault()
student.setdefault("skills", ["Python", "SQL"])

print("After setdefault():", student)

# 7. pop()
removed = student.pop("city")

print("Removed value:", removed)
print("After pop():", student)

# 8. popitem()
removed_item = student.popitem()

print("Removed item:", removed_item)
print("After popitem():", student)

# 9. copy()
new_student = student.copy()

print("Copied Dictionary:", new_student)

# 10. fromkeys()
keys = ["name", "age", "city"]

new_dict = dict.fromkeys(keys, "Unknown")

print("Dictionary using fromkeys():", new_dict)

# 11. clear()
student.clear()

print("After clear():", student)
```
